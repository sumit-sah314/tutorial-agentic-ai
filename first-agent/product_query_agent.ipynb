```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent

load_dotenv()
```

```python
PRODUCTS ={
    "laptop": {"name": "Laptop", "price": 75000, "category": "Electronics", "stock": 12},
    "smartphone": {"name": "Smartphone", "price": 25000, "category": "Electronics", "stock": 30},
    "headphones": {"name": "Headphones", "price": 3000, "category": "Accessories", "stock": 50},
    "office chair": {"name": "Office Chair", "price": 8500, "category": "Furniture", "stock": 8},
    "water bottle": {"name": "Water Bottle", "price": 500, "category": "Kitchen", "stock": 100},
}

RATINGS = {
    "laptop": {
        "rating": 4.5,
        "num_reviews": 1284,
        "reviews": ["Fast and lightweight, great for work.", "Battery lasts all day.", "Gets a bit warm under heavy load."],
    },
    "smartphone": {
        "rating": 4.2,
        "num_reviews": 3510,
        "reviews": ["Excellent camera for the price.", "Smooth display.", "Charging could be faster."],
    },
    "headphones": {
        "rating": 4.0,
        "num_reviews": 862,
        "reviews": ["Good noise cancellation.", "Comfortable for long use.", "Bass is a little weak."],
    },
    "office chair": {
        "rating": 3.8,
        "num_reviews": 417,
        "reviews": ["Good back support.", "Easy to assemble.", "Cushion flattens after a few months."],
    },
    "water bottle": {
        "rating": 4.7,
        "num_reviews": 2096,
        "reviews": ["Keeps water cold for hours.", "Doesn't leak.", "Sturdy build."],
    },
}

@tool
def get_product(product_name: str) -> dict | str:
    """Fetch a product's details (price, category, stock) by its name."""
    product = PRODUCTS.get(product_name.strip().lower())
    if product is None:
        return f"Product '{product_name}' not found. Available products: {', '.join(PRODUCTS)}"
    return str(product)

@tool
def get_product_ratings(product_name: str) -> str:
    """Fetch a product's average rating (out of 5), total number of reviews, and sample customer reviews by its name."""
    rating = RATINGS.get(product_name.strip().lower())
    if rating is None:
        return f"No ratings found for '{product_name}'. Available products: {', '.join(RATINGS)}"
    return str(rating)

print(get_product.invoke("water bottle"))
print(get_product_ratings.invoke("water bottle"))
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

agent = create_agent(llm, 
                     tools=[get_product, get_product_ratings],
                     system_prompt="You're a helpful product assistant for an online tech store.")

question = "Why people buy the water bottle"

def ask(question: str):
    result = agent.invoke({"messages": [{"role":"user", "content":question}]})
    print(result["messages"][-1].content)

ask(question)
```
