```python
import chromadb

client = chromadb.Client()
```

```python
collection = client.create_collection("my_collection")
```

```python
# Add documents to the collection. Chroma embeds them automatically by default.
collection.add(
    documents=[
        "This is a document about machine learning and neural networks.",
        "This is a document about baking sourdough bread and making delicious pastries.",
        "This is a document about traveling to Japan and exploring Tokyo."
    ],
    ids=["id1", "id2", "id3"]
)

# Query the collection for semantically similar documents (fixing the typo 'metadata' -> 'metadatas')
results = collection.get(include=["documents", "embeddings", "metadatas"])

# Display the results
for i, emb in enumerate(results["embeddings"]):
  print(f"\nDocument {i+1}")
  print("Text:", results["documents"][i])
  print("Embedding length: ", len(emb))
  print("First 10 values:", emb[:10])
```

```python
# Querying about a topic not directly mentioned ('OpenAI') to see semantic search matching
query_results = collection.query(
    query_texts=["tell me about OpenAI developments"],
    n_results=2
)

# Display the results
display(query_results)
```
